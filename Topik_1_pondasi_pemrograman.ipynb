```python
#SISTEM PROFIL DAN ZONA LATIHAN ATLET

#1. PROSES INPUT DATA (VARIABEL, TIPE DATA, & INTERAKTIF)

print("====Masukkan Data Atlet====")

#Menggunakan input untuk memasukkan nama, usia, tinggi, berat badan, & cabang olahraga

    #string khusus digunakan untuk tipe data berupa teks
nama = str(input("Nama              : "))
    #integer khusus digunakan untuk tipe data berupa bilangan bulat (digunakan karena umur berupa bilangan bulat)
usia = int(input("Usia tahun        : "))
    #float khusus digunakan untuk tipe data berupa bilangan desimal (digunakan karena tinggi dan bb adalah bilangan desimal)
tinggicm = float(input("Tinggi Badan (cm) : "))
beratkg = float(input("Berat Badan (kg)  : "))
print("Pilihan           : (Striker / Grappler)")
    #string khusus digunakan untuk tipe data berupa teks
style = str(input("Gaya Bertarung    : ")).lower()

    #("\n") digunakan untuk memberi jarak
print("\n")
print("\n")
    #("-" * 50) bermaksud untuk mencetak - sebanyak 50 kali
print("-" * 50)
print("====Masukkan 4 Sesi Data Latihan====")
print("-" * 50)

if style == 'striker':
    print("Saran: Pukulan, Tendangan, Atletik, Weightlifting")
else:
    print("Saran: Takedown, Submission, Atletik, Weightlifting")

print("-" * 50)

    #Buat list kosong sesi latihan untuk menyimpan data latihan
latihan = []

    #Gunakan 'for' untuk membuat kode input berulang
for i in range(1, 5):
    #Start 1 sampai dengan variabel ke 5
    #Range(1, 5) dimaksudkan agar kode input berulang sebanyak 4 kali (1, 2, 3, 4)

    print(f"\nSesi {i}: ")
    jenis = str(input("   Jenis Latihan          : "))
    durasi = int(input("   Durasi (menit)         : "))
    kalori = int(input("   Kalori Terbakar (kkal) : "))
    hr = float(input("   Detak Jantung (bpm)    : "))

    teval = "-"
    if jenis in ["tendangan" ,"pukulan", "takedown", "submission" ]:
        jumlah = int(input("   Jumlah Gerakan         : "))
        if durasi > 0:
            rgerakan = jumlah / durasi
        else:
            rgerakan = 0
        teval = f"{rgerakan:.1f} x/min"
    elif jenis == "atletik":
        jarak = float(input("   Jarak (km)             : "))
        if jarak > 0:
            pace = durasi / jarak
        else:
            pace = 0
        teval = f"{pace:.2f} min/km"

    #Simpan setiap data sesi dalam bentuk dictionary, lalu masukkan ke dalam list latihan
    latihan.append({'jenis': jenis, 'durasi': durasi, 'kalori': kalori, 'hr': hr, 'teval': teval})
print("-" * 50)

#2. PROSES BEKERJA

    #Hitung IMT (ubah tinggi satuan cm ke dalam m)
tinggim = tinggicm / 100
imt = beratkg / (tinggim ** 2)

    #Menggunakan percabangan if, elif, else untuk menentukan kondisi imt atlet
if imt < 18.5:
    kimt = "Underweight"
elif imt <= 24.9:
    kimt = "Normal"
elif imt <= 29.9:
    kimt = "Overweight"
else:
    kimt = "Obesitas"

    #Hitung berat(kg) / tinggi(cm) dan menentukan status ideal
bt = beratkg / tinggicm
if 0.38 <= bt <= 0.42:
    sbt = "Ideal ✅"
else:
    sbt = "Tidak Ideal❎"

    #Hitung detak jantung maksimum(HRMax) & zona latihan
hrmax = 220 - usia
    #Hitung zona latihan menggunakan persentase (50%, 70%, 85%) ---> ubah dalam bentuk desimal
zonar = hrmax * 0.50
zonas = hrmax * 0.70
zonab = hrmax * 0.85

    #Hitung total durasi latihan
tdurasi = sum(sesi['durasi'] for sesi in latihan)

    #Hitung rata-rata kalori
rkalori = sum(sesi['kalori'] for sesi in latihan) / len(latihan)

    #Cari nilai tertinggi hr dari tiap sesi latihan
hrt = max(sesi['hr'] for sesi in latihan)

#3. PRINT MENGGUNAKAN F STRING

print("\n\n")
print("=" * 26)
print("     •OUTPUT PROGRAM•     ")
print("=" * 26)
print("\n")
print("-" * 50)
print("======PROFIL DAN ZONA LATIHAN ATLET======")
print("-" * 50)

print(f"Nama             : {nama}")
print(f"Usia             : {usia} tahun")
    #:.0f bermaksud menghapus angka dibelakang koma menjadi 0 (tidak ada angka)
print(f"Tinggi Badan     : {tinggicm} cm")
print(f"Berat Badan      : {beratkg} kg")
print(f"Gaya Bertarung   : {style}")
print("-" * 50)

    #:.2f bermaksud untuk membuat angka dibelakang koma menjadi 2 angka saja
print(f"IMT              : {imt:.2f} ---> {kimt}")
print(f"Rasio B/T        : {bt:.2f} ---> {sbt}")
print(f"HRMax            : {hrmax} bpm")
print("Zona Latihan     :")

    #:.1f bermaksud untuk membuat angka dibelakang koma menjadi 1 angka saja
print("Zona Latihan     :")
print(f"    -Ringan (50%): {zonar:.1f} bpm")
print("      ==> Saran: Gunakan untuk recovery, pemanasan, atau drill teknik dasar (misal: shadow boxing ringan).")
print(f"    -Sedang (70%): {zonas:.1f} bpm")
print("      ==> Saran: Fokus untuk ketahanan kardio, pembakaran lemak, atau sparring/rolling ringan.")
print(f"    -Berat  (85%): {zonab:.1f} bpm")
print("      ==> Saran: Peningkatan VO2Max, kecepatan eksplosif, angkat beban berat, atau sparring intens.")
print("\n")

print("\n")
print("-" * 50)
print("======RINGKASAN LATIHAN HARIAN======")
print("-" * 50)
    #{teks:<angka} berarti teks rata kiri dengan lebar spasi tertentu
print(f"{'No':<2} | {'Jenis Latihan':<13} | {'Durasi':<8} | {'Kalori':<6} | {'HR':<4} | {'Evaluasi'}")
print("-" * 50)

    #Menggunakan enumerate untuk membuat (i) mulai dari 1
for i, sesi in enumerate(latihan, start=1):
    print(f"{i:<2} | {sesi['jenis']:<13} | {sesi['durasi']:<6} m | {sesi['kalori']:<6} | {sesi['hr']:<4.0f} | {sesi['teval']}")

print("-" * 50)
print(f"Total Durasi     : {tdurasi} menit")
print(f"Rata-rata Kalori : {rkalori:.2f} kkal")
print(f"HR Tertinggi     : {hrt} bpm")
print("-" * 50)
```
